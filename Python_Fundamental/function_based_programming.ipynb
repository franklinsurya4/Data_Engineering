```python
def cleanup(dir:str):
    try:
        import os
        os.rmdir(dir)#empty dir
    except Exception as e:
        import shutil
        # shutil is a build-in module in python used for high-level file and directory operations, such as copying, moving, and deleting folders.
        shutil.rmtree(dir,ignore_errors=True)#directory with some subfiles/subdirs
        return "Non Empty Dir Deleted Successfully"
    else:
        return "Empty Dir Deleted Successfully"
return_result=cleanup("D:\\dataset\\tmp")
print(return_result)
```

```python
#2. Calling Functions - Positional & Keyword/named Arguments
#AI says, Parameter (named) = Placeholder, Argument(positition) = actual value
def calculate_offer(amt,off_pct,shipping_charges):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return total_bill_amt

#Positional Argument Function Call
result=calculate_offer(1000,.05,100)#positional arguments (arguments)#Internally amt=1000, off_pct=.05, shipping_charges=100
print('Positional with right mapping',result)
result=calculate_offer(1000,100,.05)#Internally amt=1000, off_pct=100, shipping_charges=.05
print('Positional with wrong mapping',result)#wrong result if i use postitional argument
#Can we relate with database queries?
#insert into tablename values(1,'irfan',44);#positional provided the input is as per the order the table columns
#select * from tablename;#positional


#Keyword/Named Argument Function Call
result=calculate_offer(amt=1000,off_pct=.05,shipping_charges=100)
print('Named is always right in mapping',result)
result=calculate_offer(amt=1000,shipping_charges=100,off_pct=.05)
print('Named is always right in mapping',result)#Right result if i use named/keyword parameter

#If we pass less or more params than expected
#more number of params/arguments fails
#result_more_arguments=calculate_offer(1000,.05,100,0)#errored out
#result_more_params=calculate_offer(amt=1000,off_pct=.05,shipping_charges=100,unknown=0)

#less number of params/arguments
#result_less_arguments=calculate_offer(1000,.05)#errored out
#result_less_arguments=calculate_offer(amt=1000,off_pct=.05)#errored out

#insert into tablename(id,age,name) values(1,44,'irfan');#named notation
#select name,age,id from tablename;#named


#If I have chance of passing less number of arguments?
#We can go for DEFAULT ARGUMENT FUNCTIONS, Rule: default arguments has to be in the last
def calculate_offer(amt,off_pct=.0,shipping_charges=0):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return round(total_bill_amt)
#create table tablename(id int,name varchar(100),age int default 0);
#insert into tablename(id,name) values(1,'irfan');#named notation with age=0 by default inserted

#less number of params/arguments
result_less_arguments=calculate_offer(1000)#Default arguments works with positional..
print(result_less_arguments)

#We are using positional, named & default
def calculate_offer(amt,shipping_charges,off_pct=.0):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return round(total_bill_amt)

result_less_arguments=calculate_offer(1000,shipping_charges=100)#Default arguments works..
print(result_less_arguments)

#result_less_arguments=calculate_offer(amt=1000,off_pct=.05)#Default arguments works..
```

```python
#Usecase 1:Arbitrary Argument Function - Weather Station Stats - Arbitrary arguments (*args)

#You are building software for a weather station. Throughout the day, temperature sensors send a batch of readings back to your system. Because the number of active sensors changes depending on the location, your function must accept any number of numerical readings and calculate key summary statistics.

sensor_7am=22,21,22,20,23,25,27,28
sensor_10am=22,21,22,20,23,25,27,28,41,30,45

import statistics as st

print(max(sensor_7am))
print(min(sensor_7am))

print(st.median(sensor_7am))
print(st.mean(sensor_7am))

def weather_stats(*args):
    import statistics as st
    print(args)

    max_temp=max(args)
    min_temp=min(args)
    mean_temp=st.mean(args)#sum(args)/len(args)
    median_temp=st.median(args)
    std_dev_temp=st.stdev(args)

    return {'max_temp':max_temp,'min_temp':min_temp,'median_temp':median_temp,'mean_temp':mean_temp,'std_dev_temp':std_dev_temp}
print(weather_stats(22,21,22,20,23,25,27,28))
print(weather_stats(22,21,22,20,23,25,27,28,41,30,45))
```

```python
#Usecase 2: IT Gross Sal Calculator - *args
#CTS -> sal, bon, inc (100000,5000,.02)
#INFY -> sal, bon (100000,5000)
#Can i achieve using arb arg fun? yes, or which one is better? KWArgs is better because we have to handle arguments using name, not by using position.

def hr_workways_sal_fun(*args):
    print(args)
    
    if args[0]=='CTS':
        sal=args[1]
        bon=args[2]
        inc=args[3]
        inc_amt=sal*inc
        final_sal=sal+bon+inc_amt
    elif args[0]=='INFY':
        sal=args[1]
        bon=args[2]
        final_sal=sal+bon
    elif args[0]=='HCL':
        sal=args[1]
        final_sal=sal
    else:
        return 0
    return int(final_sal)

#Emp1
final_sal_cts=hr_workways_sal_fun('CTS',100000,5000,.02)
final_sal_infy=hr_workways_sal_fun('INFY',100000,5000)
final_sal_hcl=hr_workways_sal_fun('HCL',100000)
final_sal_wipro=hr_workways_sal_fun('WIPRO',100000,6000)

print("cts final sal",final_sal_cts)
print("infy final sal",final_sal_infy)
print("hcl final sal",final_sal_hcl)
print("wipro final sal",final_sal_wipro)
```

```python
#Usecase 2: IT Gross Sal Calculator - **kwargs
#CTS -> sal, bon, inc (100000,5000,.02)
#INFY -> sal, bon (100000,5000)
#Can i achieve using arb arg fun? yes, or which one is better? KWArgs is better because we have to handle arguments using name, not by using position.

def hr_workways_sal_fun(**kwargs):
    print(kwargs)
    compname=kwargs['compname']
    if compname in ['CTS','INFY','HCL']:
        sal=kwargs['sal']
        bon=kwargs.get('bon',0)
        inc=kwargs.get('inc',0)
        inc_amt=sal*inc
        return int(sal+bon+inc_amt)

#Emp1
final_sal_cts=hr_workways_sal_fun(compname='CTS',sal=100000,bon=5000,inc=.02)
final_sal_infy=hr_workways_sal_fun(compname='INFY',sal=100000,bon=5000)
final_sal_hcl=hr_workways_sal_fun(compname='HCL',sal=100000,inc=.03)
final_sal_wipro=hr_workways_sal_fun(compname='WIPRO',sal=100000,bon=6000)

print("cts final sal",final_sal_cts)
print("infy final sal",final_sal_infy)
print("hcl final sal",final_sal_hcl)
print("wipro final sal",final_sal_wipro)
```

```python
#6. Higher Order Functions (HOF)
#final_sal=sal+bon
def sqrt(x):
    return x*x

def fun1(a,b):#HOF (Func passed as an input to another function)
    return b(a)

print(fun1(10,sqrt))
```

```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    return a/b

def fun1(a,b,xyz):#We can pass value or function as an input to a function
    result=xyz(a,b)
    return result

print(fun1(10,5,add))
print(fun1(10,5,sub))
print(fun1(10,5,mul))
print(fun1(10,5,div))
```

```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def fun1(a,b,xyz):#If we don't use HOF, then reusability is not fully achieved...
    if xyz=='add':
        result=a+b
    elif xyz=='sub':
        result=a-b
    elif xyz=='mul':
        result=a*b
    return result

print(fun1(10,5,'add'))
print(fun1(10,5,'sub'))
print(fun1(10,5,'mul'))
print(fun1(10,5,'div'))
```
