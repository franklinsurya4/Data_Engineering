```python
'''usecase2: Swiggy food delivery usecase

Pseudo code (prompt - context):

Identify the variables - min_cart_amt, disc_pct_whole, max_disc_amt

Identify the input (runtime) - cart_value

First check: if cart_value>=min_cart_amt then i am going to apply offer else i am going notify user to add some more amout to get offer. min_cart_amt - cart_value

Calculate discount in percentage (fraction): disc_pct=10/100

Calculate discount amount after applying to the cart amt : if the discount amt > max discount, then reduce max discount to the cart amount and produce final amount, otherwise reduce calculated discount amt to the cart amount and produce final amount.

eg. if customer add cart of 1200 , check 1200>600, 1200*.10 -> 120 then check 120>=100 then give discount 1200-100 & result is 1100.'''
```

```python
min_cart_amt=600
disc_pct_whole=10
max_disc_amt=100

cart_amt=int(input("Choose the products hence amount added in the cart"))

if cart_amt>=min_cart_amt:
    print("eligible for offer")

    disc_pct_fraction=disc_pct_whole/100 # eg. 10/100 -> .10
    calc_discount_amt=cart_amt * disc_pct_fraction #eg. 800 *.10 -> 80

    if max_disc_amt >= calc_discount_amt:
        final_bill_amount=cart_amt-calc_discount_amt
        print(f"final bill amt after reducing discount of 10%: {calc_discount_amt} final bill amt {final_bill_amount}")
    else:
        final_bill_amount=cart_amt-max_disc_amt
        print(f"final bill amt after reducing discount flat discount of max_disc_amt :{max_disc_amt} final billl amt {final_bill_amount}")
else:
    print(f"not eligible for offer, add this amount in your cart {min_cart_amt-cart_amt}")
```

```python
#Implement Password validation. Keep prompting the user for a password indefinitely (simulating a do-while loop) and use a break statement to exit the loop only when the password meets all required conditions.

#What is entry controlled loop -
# If we control the execution of the loop at the time of entry itself
startval=1
endval=5
while startval<=endval:#entry controlled
    print(f"execute the loop {startval}")
    startval+=1
```

```python
#What is exit controlled loop -
# If we control the execution of the loop at the time of exit
startval=1
endval=5
while True:#dont put any control in the entry, just allow the loop to run
    if startval>endval:#exit controlled loop
        print(f"terminate the loop because {startval} > {endval}")
        break #exit controlled loop
    print(f"execute the loop {startval}")
    startval+=1


print("after break, this line of code executed")
```

```python
#Usecase4: Implement Password validation. Keep prompting the user for a password infinately (simulating a do-while loop) and use a break statement to exit the loop only when the password meets all required conditions.
actual_password='Inceptez@123'
#In python we don't have exclusively do while looping, but we can use while True
# (exit control loop to achieve do while loop)
while True:#not entry control loop
    user_password=input("enter the password")
    if user_password==actual_password:
        print("you are allowed to access the application")
        break # Exit control
    else:
        print("retry the password")
```

```python
#With an Exception handler we have control to handle the error
#Function (FBP)
while True:
    try:
        actual_password=33131
        for i in range(3):
            user_password=int(input("enter the otp"))
            if user_password==actual_password:
                print("you are allowed to access the application")
                break#Exit control
            else:
                print("retry the password")
        else:
            print("max attempts reached")
    except Exception as e:
        print(f"Ensure to give only number: {e}")
    break
    print("next block")
```
