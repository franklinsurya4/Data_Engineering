```python
#Voting Example where no exception occurs
#Conclusion of exception handling blocks:
#Try block will always executed as a first block? True
#Try block needs any one of these blocks - except block or finally block or else block? False (else alone cant work)
#Try block needs any one of these blocks - except block or finally block? True
#If Exception occurs in try then except block will be executed & else block is also executed? False (only except is executed)
#If Exception occurs in try then except block will be executed & finally block is also executed? True
#If try block is success then else block alone is executed, though i have except & finally block? False (else & finally will also execute)
#If try block is success then else block and finally block is executed? True
#else executed when no exception? True
#except executed when exception? True
#finally executed when exception or when no exception? True
try:
    age='19'
    if age>=18:
        print("Eligible for voting")
    else:
        print("Not Eligible for voting")
except Exception as err_message:#Run when try is failed
    print("Exception Block: something went wrong : {err_message}")
else:#Run when try is success
    print("Else Block: Everything went fine inside the try block")
    #os.rmdir("D:\\tmp")
finally:#Whether Run is success or failed (exception occured)
    print("Finally Block : will run at any cost")
    print("Doing some cleanup, if try is successful or failed")
    #os.rmdir("D:\\tmp")
```

```python
#Code with exception handler
try:
    a=101
    #print(b)#nameerror
    #print('10'+44)#typeerror
    #print(int('hello'))#valueerror
    lst=['Saravanan','Yuvaraj']
    #print(lst[2])#indexerror
    dict1={1:'irfan',2:'saravanan',3:'Yuvaraj'}
    #print(dict1[10])#Keyerror
    #print(100/0)#ZeroDivisionError
    assert a==100,"ensure a is assigned as 100"
except NameError as err:
    print("Print the variable after defining it:",err)
except TypeError as err:
    print("Python is strongly typed language:",err)
    print(int('10')+44)
except ValueError as err:
    print("Alphabets can't be converted to int:",err)
    print(int('10')+44)
except IndexError as err:
    print("Index error :",err)
    print(lst[len(lst)-1])
except KeyError as err:
    print("Dictionary key is not found :",err)
    print("try to use only these keys:",dict1.keys())
except ZeroDivisionError as err:
    print("Denominator should'nt be zero:",err)
except AssertionError as err:
    print("Assertion error",err)
    a=100
    print(a)
except Exception as err:
    print("Some error occured",err)
```

```python
#Custom Exception Handler to handle
#If the predefined exceptions of the list provided is not enough to handle some specific exceptions that we want to customize. If I want to raise an exception in case it is not provided already, then we can create custom exception.
class InceptezError(Exception): #Inheritance (OOPs)
    pass
try:
    aspirants_joined=input("Enter Aspirant name:")
    aspirants_list=["Preethi","Sujithra","Nathiya","Uvaraj"]
    if aspirants_joined in aspirants_list:
        print("He/She is a registered Aspirant")
    else:
        raise InceptezError(f"You are not a registered Aspirant {aspirants_joined}")
except InceptezError  as errmsg:
    print("Error Message:",errmsg)
    print(f"Remove this aspirant from zoom meeting {aspirants_joined}")
```

```python
def clean(waittime):
        try:#main block
                import time
                print(f"logic to do clean")
                time.sleep(100/waittime)#If i pass 0 as denominator, error occurs & program terminated , moving the control from line 4 to line
                print("Cleaning completed")
                return True
        except Exception as ex_msg:
            print(f"cleaning failed due to : {ex_msg}, some other error")
clean('10')
```

```python
def clean(waittime):
        try:#main block
                import time1
                print(f"logic to do clean")
                time.sleep(100/waittime)#If i pass 0 as denominator, error occurs & program terminated , moving the control from line 4 to line
                print("Cleaning completed")
                return True
        except ZeroDivisionError as ex_msg:#handler block
            print(f"cleaning failed due to : {ex_msg} , don't enter zero as denominator")
            return False
        except TypeError as ex_msg:#handler block
            print(f"cleaning failed due to : {ex_msg}, enter only number")
            return False
        except Exception as ex_msg:
            print(f"cleaning failed due to : {ex_msg}, some other error")
clean('10')
```
